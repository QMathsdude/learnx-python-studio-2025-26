## We will start with normal pairwise data plotting

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)
y1 = np.cos(x)
rand = np.random.rand(100)


fig, axs = plt.subplot_mosaic([['A', 'B', 'C'],
                               ['D', 'E', 'F'],
                               ['.', 'G', '.']], figsize=(20, 5))
axs['A'].plot(x,y, linewidth=2, color='red')
axs['B'].scatter(x, rand, s=10, color='green')
axs['C'].hist(rand, bins=20, width=0.04)
axs['D'].stem(x, y)
axs['E'].fill_between(x, y, alpha=0.5, linewidth=2, color='red', facecolor='blue')
axs['F'].stackplot(x, y, y1)
axs['G'].stairs(rand, color='black', fill=True, alpha=0.5)
```

## Then there is also statistical distribution



```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.random.rand(100)

fig, axs = plt.subplot_mosaic([['A', 'B', 'C'],
                               ['D', 'E', 'F'],
                               ['G', '.', 'H']], figsize=(20, 5))

axs['A'].hist(y, bins=20, width=0.05, color='blue', alpha=0.5, label='Histogram')
axs['B'].boxplot(y, vert=False, patch_artist= True, boxprops=dict(facecolor='orange', color='red'), medianprops=dict(color='green'))
axs['C'].errorbar(x, y, yerr=0.1, fmt='o', color='blue', ecolor='red', elinewidth=2, capsize=4)
axs['D'].eventplot(y, color='green')
axs['E'].hist2d(x, y, bins=20, cmap='Blues')
axs['F'].hexbin(x, y, gridsize=20, cmap='Greens')
axs['G'].pie([40,20,40], colors=['cyan', 'white', 'green'], autopct='%.1f%%')
axs['H'].ecdf(y, color='red', linewidth=2)
```

## Finally there the Gridded Data



```python
import matplotlib.pyplot as plt
import numpy as np

X, Y = np.meshgrid(np.linspace(-3, 3, 16), np.linspace(-3, 3, 16))
Z = (1 - X/2 + X**5 + Y**3) * np.exp(-X**2 - Y**2)
V = np.diff(Z[1:, :], axis=1)
U = -np.diff(Z[:, 1:], axis=0)

fig, axs = plt.subplot_mosaic([['A', 'B', 'C'],
                               ['D', '.', 'E']], figsize=(20, 5))

axs['A'].imshow(Z, cmap='viridis')
axs['B'].pcolormesh(X, Y, Z, shading='auto', cmap='plasma')
axs['C'].contour(X, Y, Z, levels=10, colors='black')
axs['D'].contourf(X, Y, Z, levels=20, cmap='inferno')
axs['E'].streamplot(X[:-1, :-1], Y[:-1, :-1], U, V, color='blue')
```

## Patches
### But what if i want have text or shapes on the plot you made

That where the class matplotlib.artist and .patches come in

Artist and patches are used to make custom element in a plt without interferring with the data in it.

we can put texts,boxes, arrows, annotations and many more

theres a lot of feature but we will only cover the important functions

looks though this website to find all the other uses

[For all features click here](https://matplotlib.org/stable/api/patches_api.html)



```python
import matplotlib.pyplot as plt
import matplotlib.patches as pc
import numpy as np

x = np.linspace(0, 10, 100)
y = np.exp(x)

fig = plt.figure(figsize=(20,5))

ax1 = plt.subplot2grid((1, 2), (0, 0), facecolor='orange')

ax1.plot(x, y, linestyle='--', color='black')

## add annotation to the plot
ax1.annotate('look here',
            xy=(8, np.exp(8)),
            xytext=(0.1, 0.5),
            textcoords='figure fraction',
            bbox=dict(boxstyle="round4,pad=.5"),
            arrowprops=dict(arrowstyle="->",connectionstyle="arc3,rad=.2"))

## Create a circle
circle = pc.Circle((0, 0), radius= 1, edgecolor='red', linewidth=2)
cax = plt.axes([0.3, 0.5, 0.5, 0.4])
cax.set_xlim(-2, 2)
cax.set_ylim(-2, 2)
cax.add_patch(circle)
cax.set_aspect('equal')
cax.set_axis_off()
cax.text(0, 0, 'Circle', ha='center', va='center', fontsize=12, color='red')

## Create an ellipse
ellips = pc.Annulus((0, 0), r=(1,1.5), width = 0.5, edgecolor='red', linewidth=2)
eax = plt.axes([0.3, 0.2, 0.5, 0.4])
eax.set_xlim(-2, 2)
eax.set_ylim(-2, 2)
eax.add_patch(ellips)
eax.set_aspect('equal')
eax.set_axis_off()

## Create an arrow
Arrow = pc.Arrow(0, 0, 1, 1, width=0.5, edgecolor='red', facecolor='red')
aax = plt.axes([0, 0.2, 0.3, 0.4])
aax.set_xlim(-2, 2)
aax.set_ylim(-2, 2)
aax.add_patch(Arrow)
aax.set_aspect('equal')
aax.set_axis_off()


## Create boxes for you elements
frame_ax = plt.axes([0.5, 0.55, 0.15, 0.3])
frame_ax.patch.set_alpha(0.5)
frame_ax.spines['top'].set(color='burlywood',linewidth = 2)
frame_ax.spines['right'].set(color='burlywood',linewidth = 2)
frame_ax.spines['bottom'].set(color='burlywood',linewidth = 2)
frame_ax.spines['left'].set(color='burlywood',linewidth = 2)
frame_ax.set_xticks([])
frame_ax.set_yticks([])
```

## Widgets

widgets can be used to make interactable elements that will update our graph
this include:
<div style="padding: 10px; border: 2px solid ; border-radius: 5 px;">

- slider
- button
- radio button
- input box
</div>

### When we make interactable elements we must follow a certain format to make sure the updates work properly

1. initialise everything (figures, axis, and create the slider)
2. update (update the values that the widgets control)
3. plot (use the updated value to create a new plot)



```python
### remember each step must be done in different functions 
### import as normal
import matplotlib.pyplot as plt
from matplotlib.widgets import Button, Slider, TextBox, RadioButtons
import numpy as np

### create a class to handle the interactive plot

class InteractivePlot:
    ### initialize the plot and widgets
    def initialize(self):
        ### our initial data
        self.w = 0
        self.x = np.linspace(0, 10, 100)
        self.y = np.sin(self.x*np.radians(self.w))
        ### create the figure and axis
        self.fig = plt.figure(figsize=(10,5))
        self.ax = plt.subplot2grid((1, 1), (0, 0))
        plt.subplots_adjust(left=0.05, right= 0.95, bottom=0.32, top=0.92, hspace=0.4)
        ### create the slider
        self.slider_ax = plt.axes([0.1, 0.2, 0.5, 0.03])
        self.slider = Slider(self.slider_ax, 'Frequency', 0, 360, valinit=180)
        ### when the sldier actually moves it will run the update function
        self.slider.on_changed(self.update)
        ### we make sure figure is plotted before we can update it
        self.plot()  
        
    ### then we update the value of w based on the slider value
    def update(self, w):
        self.w = self.slider.val
        ### we make sure the plot is updated with the new value of w
        self.plot()
        
    ### finally we create a function to plot the data
    
    def plot(self):
        ### we clear the axis before plotting the new data
        self.ax.clear()
        self.x = np.linspace(0, 10, 100)
        self.y = np.sin(self.x*np.radians(self.w))
        self.ax.plot(self.x, self.y, color='blue', linewidth=2)
        self.fig.canvas.draw_idle()

### now the initialization funtion happens everytime we run the code
InteractivePlot().initialize()
plt.show()
```
